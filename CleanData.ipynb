```python
data=pd.read_csv("Project 1/olist_products_dataset.csv")
```

```python
df=pd.DataFrame(data)
print(df)
```

```python
print("--- SHAPE ---")
print(df.shape)

print("\n--- INFO ---")
print(df.info())

print("\n --- MISSING VALUE ---")
print(df.isnull().sum())

print("\n --- DUPLICATES ---")
print(df.duplicated().sum())

print("\n --- UNIQUE VALUES ---")
print(df.nunique())
```

```python
# 1. fill the missing value og product_category_name
df['product_category_name'] = df['product_category_name'].fillna('Unknown')

# 2. fill the missing value of metadata
df['product_name_lenght'] = df['product_name_lenght'].fillna(0)
df['product_description_lenght'] = df['product_description_lenght'].fillna(0)
df['product_photos_qty'] = df['product_photos_qty'].fillna(0)

# 3. fill the missing value of product weight,length,height and width
dim_cols = ['product_weight_g', 'product_length_cm', 'product_height_cm', 'product_width_cm']
for col in dim_cols:
    df[col] = df[col].fillna(df[col].median())
```

```python
print(df.isnull().sum())
```

```python
df.to_csv("Project 1/clean_olist_products_dataset.csv")
```

```python
import pandas as pd

# Load the cleaned files 
orders = pd.read_csv("Project 1/clean_olist_order_dataset.csv")
order_items = pd.read_csv("Project 1/olist_order_items_dataset.csv")
products = pd.read_csv("Project 1/clean_olist_products_dataset.csv")
translation = pd.read_csv("Project 1/clean_product_category_name_translation.csv")
payments = pd.read_csv("Project 1/olist_order_payments_dataset.csv")
customers = pd.read_csv("Project 1/clean_olist_customers_dataset.csv")

# 2. Step-by-step merge 
df_merged = orders.merge(order_items, on='order_id', how='left', suffixes=('_orders', '_items'))
df_merged = df_merged.merge(products, on='product_id', how='left', suffixes=('', '_prod'))
df_merged = df_merged.merge(translation, on='product_category_name', how='left', suffixes=('', '_trans'))
df_merged = df_merged.merge(payments, on='order_id', how='left', suffixes=('', '_pay'))
df_merged = df_merged.merge(customers, on='customer_id', how='left', suffixes=('', '_cust'))

print("Merge Complete! Merged Shape:", df_merged.shape)
```

## Which product category has the highest revenue

```python
#category wise total revenue
top_category=(
    df_merged.groupby('product_category_name_english')['payment_value'].sum().sort_values(ascending=False).reset_index()
)

print("Top 5 Revenue Categories:")
print(top_category.head())
```

## Which month had peak sales

```python
# Ensure the purchase timestamp in datetime 
df_merged['order_purchase_timestamp'] = pd.to_datetime(df_merged['order_purchase_timestamp'])

# extract year-month
df_merged['year_month'] = df_merged['order_purchase_timestamp'].dt.to_period('M')

# Monthly revenue 
monthly_sales = (
    df_merged.groupby('year_month')['payment_value']
    .sum()
    .sort_values(ascending=False)
    .reset_index()
)

print("Peak Sales Month:")
print(monthly_sales.head(1))
```

## Which Region Performs Best

```python
# Customer State-wise revenue
top_regions = (
    df_merged.groupby('customer_state')['payment_value']
    .sum()
    .sort_values(ascending=False)
    .reset_index()
)

print("Top 5 Performing States:")
print(top_regions.head())
```

## What is the average order value (AOV) trend

```python
# Total Revenue per month / Total Distinct Orders per month
aov_trend = (
    df_merged.groupby('year_month')
    .agg(total_revenue=('payment_value', 'sum'), total_orders=('order_id', 'nunique'))
    .reset_index()
)

aov_trend['AOV'] = aov_trend['total_revenue'] / aov_trend['total_orders']

print("Monthly AOV Trend:")
print(aov_trend.head(10))
```

## What is the customer review score distribution

```python
reviews = pd.read_csv("Project 1/olist_order_reviews_dataset.csv")

# Review score distribution (%)
review_dist = (reviews['review_score'].value_counts(normalize=True) * 100).reset_index()
review_dist.columns = ['review_score', 'percentage']

print("Review Score Distribution (%):")
print(review_dist.sort_values(by='review_score', ascending=False))
```
